#### Структура файла `FONT.BIN`:
- `0x0000 - 0x001F`: Пусто (32 байта нулей)
- `0x0020 - 0x035F`: Английский алфавит 16x16 (832 байт) - A-Z (832 / 32 = 26 символов)
- `0x0360 - 0x061F`: Пусто (704 байта)
- `0x0620 - 0x095F`: Английский алфавит 8x16 (832 байта) 
- `0x0960 - 0x0A7F`: Пусто (288 байт)
- `0x0A80 - 0x`
- `0x0C10 - ... `: Следующие блоки

#### Пытаюсь разобрать область `0x0020 - 0x035F`

```python
import os
from PIL import Image, ImageDraw
from IPython.display import display

with open('FONT_BIN/Data/FONT.BIN', 'rb') as f:
    data = f.read()

start = 0x020
char_size = 32
char_width = 16
char_height = 16
    
# Создаем изображение 256x16 (32 символа * 8 пикселей = 256)
img = Image.new('1', (416, 16), 1)  # 1 - белый фон (mode '1' - черно-белое)
pixels = img.load()    

for char in range(26):  # 0-25
    
    row_in_image = 0  
    col_in_image = char * char_width
    
    for row in range(char_height):  
        pos = start + char * char_size + row * 2
        if pos + 1 >= len(data):
            break
            
        byte1 = data[pos]  
        byte2 = data[pos + 1]
        
        for col in range(8):
            if byte1 & (1 << (7-col)):  # Если бит установлен
                x = col_in_image + col
                y = row_in_image + row
                if x < 416 and y < 32:  # Проверяем границы изображения
                    pixels[x, y] = 0  # 0 - черный цвет
                    
        for col in range(8):
            if byte2 & (1 << (7-col)):  # Если бит установлен
                x = col_in_image + col + 8
                y = row_in_image + row
                if x < 416 and y < 32:  # Проверяем границы изображения
                    pixels[x, y] = 0  # 0 - черный цвет
        

img.save('Result/font_output_0x0020-0x035F.png')

new_width = img.width * 5
new_height = img.height * 5
img_resized = img.resize((new_width, new_height), Image.Resampling.NEAREST)  # NEAREST лучше для пиксельной графики

display(img_resized)
```

#### Пытаюсь разобрать весь файл

```python
import os
from PIL import Image, ImageDraw
from IPython.display import display
import math

with open('FONT_BIN/Data/FONT.BIN', 'rb') as f:
    data = f.read()

start = 0x000
char_size = 32
char_width = 16
char_height = 16

# Для 256 символов рассчитываем оптимальную сетку
total_chars = 256
chars_per_row = int(math.sqrt(total_chars))  # 16 символов в строке
chars_per_col = total_chars // chars_per_row  # 16 символов в столбце

# Масштабируем размеры символов, чтобы заполнить 2048x2048
scaled_width = 2048 // chars_per_row  # 2048/16 = 128 пикселей на символ
scaled_height = 2048 // chars_per_col  # 2048/16 = 128 пикселей на символ

# Создаем изображение 2048x2048
img = Image.new('1', (2048, 2048), 1)
pixels = img.load()    

for char in range(total_chars):
    # Вычисляем позицию символа в сетке
    row_in_image = (char // chars_per_row) * scaled_height
    col_in_image = (char % chars_per_row) * scaled_width
    
    for row in range(char_height):  
        pos = start + char * char_size + row * 2
        if pos + 1 >= len(data):
            break
            
        byte1 = data[pos]  
        byte2 = data[pos + 1]
        
        # Масштабируем символ до нужного размера
        for col in range(char_width):
            # Определяем, какой байт использовать
            if col < 8:
                byte = byte1
                bit = 7 - col
            else:
                byte = byte2
                bit = 7 - (col - 8)
            
            if byte & (1 << bit):
                # Масштабируем пиксель до scaled_width x scaled_height
                for sy in range(scaled_height // char_height):
                    for sx in range(scaled_width // char_width):
                        x = col_in_image + col * (scaled_width // char_width) + sx
                        y = row_in_image + row * (scaled_height // char_height) + sy
                        if x < 2048 and y < 2048:
                            pixels[x, y] = 0

img.save('Result/font_output.png')

new_width = img.width // 4
new_height = img.height // 4
img_resized = img.resize((new_width, new_height), Image.Resampling.NEAREST)

display(img_resized)
```
