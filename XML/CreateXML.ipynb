Создания XML файла для замены кадров в файле MV1C.STR 

```python
with open('replace_str.xml', 'w') as f:
    f.write('<?xml version="1.0"?>\n')
    f.write('<str-replace version="0.3">\n')
    for i in range(2090):  # от 0 до 2089
        f.write(f'    <replace frame="{i}">STR1/MV1C.STR[0][{i:04d}].bmp</replace>\n')
    f.write('</str-replace>\n')
```
